```python
import numpy as np
from keras import Sequential
from keras.layers import Dense
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('cechy_tekstur.csv')
X = df.drop(columns='kategoria').values
y = df['kategoria'].values
print(X.shape, y.shape)
print(y[:5])
```

```python
y_int = LabelEncoder().fit_transform(y)
y_onehot = OneHotEncoder(sparse_output=False).fit_transform(y_int.reshape(-1, 1))

X_train, X_test, y_train, y_test = train_test_split(X, y_onehot, test_size = 0.3, random_state=0, stratify=y_int)

skaler = StandardScaler()
X_train = skaler.fit_transform(X_train)
X_test = skaler.transform(X_test)

print(y[:2], y_int[:2])
print(y_onehot[:2])
print(X_train.shape, X_test.shape, y_train.shape)
```

```python
model = Sequential()
model.add(Dense(10, activation='sigmoid', input_dim=72))
model.add(Dense(4, activation='softmax'))

model.compile(loss='categorical_crossentropy', optimizer='sgd', metrics=['accuracy'])
model.summary()
```

```python
model.fit(X_train, y_train, epochs=100, batch_size =10, shuffle=True)
```

```python
y_pred = model.predict(X_test)
y_test_int = np.argmax(y_test, axis=1)
y_pred_int= np.argmax(y_pred, axis=1)

print(confusion_matrix(y_test_int, y_pred_int))
print('Dokładność na zbiorze testowym:', np.mean(y_test_int == y_pred_int))
```
